```python
import pandas as pd 
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_spli
```

```python
df = pd.read_csv("std_dataset.csv")

df.head()
```

```python
# 1 input (StudyHours) => 1 output  (Marks)

x = df[["StudyHours"]]   # input 2d me hota h 
y = df["Marks"]

x.head()
y.head()
```

```python
# step 3 spliting data in train and test 

x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size = 0.2 , random_state= 42 
    
    )  # x,y,test_size = 0.2  this 0.2 decide the testing data portion means 20% data
```

```python
# step 5 model implement 

model = LinearRegression() #it is an initialisation method of oops parent child class

model.fit(x_train,y_train) # fit give the data of training to model 


```

```python
y_pred = model.predict(x_test) # input testing data 

print(y_pred)
```

```python
print("actual data ")
print(y_test)

print( "predict dagta ")
print( y_pred)
```

```python
new_result = pd.DataFrame(
    {
        "StudyHours" :[3]
    }
)

marks = model.predict(new_result)
print("new result",marks[0])
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test,y_pred)

print(r2)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test,y_pred)

print(mae)
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test,y_pred)

print(mse)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("RMSE:", rmse)
```
